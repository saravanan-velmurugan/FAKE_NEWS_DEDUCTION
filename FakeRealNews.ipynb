```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import classification_report

import joblib
```

```python
df = pd.read_csv('Fake_Real_News_Data.csv')
df
```

```python
df['sentence'] = df.title + df.text
df
```

```python
df['sentence'][6333]
```

```python
X = df.sentence
y = df.label
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(X,y,train_size=0.8,random_state=42)
```

```python
y.value_counts()
```

```python
main_pipe = Pipeline(
    steps=[
        ('vector',TfidfVectorizer()),
        ('model',LogisticRegression())
    ]
)
```

```python
main_pipe.fit(xtrain,ytrain)
```

```python
ytrain_pred = main_pipe.predict(xtrain)
ytest_pred = main_pipe.predict(xtest)
```

```python
print("Train Result......")
print(classification_report(ytrain,ytrain_pred))
print()
print("Test Result.......")
print(classification_report(ytest,ytest_pred))
```

```python
joblib.dump(main_pipe,'FakeRealNews.pkl')
```
